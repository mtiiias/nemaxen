# Proyecto Machine Learning: Análisis de Spotify
En este notebook realizaremos las primeras fases de la metodología CRISP-DM aplicadas a un dataset de Spotify, con el objetivo de preparar los datos para futuros modelos predictivos.

```python
# Importamos las librerías requeridas
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.preprocessing import MinMaxScaler

# Cargamos el dataset
df = pd.read_csv("Spotify_Tracks_Dataset.csv")
```

## Fases 1 y 2: Comprensión del Negocio y de los Datos
### Definición de Variables Objetivo (Targets)

Tras un análisis inicial de las características del dataset de Spotify y considerando el contexto del negocio, hemos definido las siguientes variables objetivo para las distintas tareas de Machine Learning:

**1. Tarea de Regresión:**
* **Target seleccionado:** `popularity` (numérica discreta/continua de 0 a 100).
* **Justificación de negocio:** Predecir la popularidad de una pista en función de sus características acústicas permite a las discográficas y a las plataformas de streaming anticipar el éxito de una canción, optimizando las campañas de marketing.

**2. Tarea de Clasificación:**
* **Target seleccionado:** `track_genre` (categórica multiclase).
* **Justificación de negocio:** Asignar automáticamente un género musical a una canción nueva basándose en sus atributos de audio es fundamental para los algoritmos de recomendación de la plataforma, facilitando la catalogación masiva.

### Análisis Exploratorio y Estadístico

Para comprender mejor la naturaleza de nuestras variables, aplicamos estadísticos de tendencia central (media, mediana) y de dispersión (desviación estándar, mínimos, máximos y cuartiles).

**Hallazgos principales:**
1. **Escalas variables:** Observamos que características acústicas como `danceability`, `energy` o `valence` están en una escala de 0 a 1, mientras que `loudness` tiene valores negativos y `tempo` alcanza más de 200 BPM. Esto indica que más adelante será imperativo aplicar técnicas de estandarización o normalización.
2. **Posibles Outliers (Valores Atípicos):** En la variable `duration_ms` observamos una gran dispersión. Mientras que la media es de ~3.8 min, existe un valor máximo de aprox. 87 minutos, lo que sugiere la presencia de anomalías que deberán ser tratadas.

```python
# Verificamos la estructura y los valores nulos iniciales
print("--- Información del Dataset ---")
df.info()
print("\n--- Conteo de Valores Nulos ---")
print(df.isnull().sum()[df.isnull().sum() > 0])

# Generamos un resumen estadístico de las variables numéricas
print("\n--- Estadísticos Descriptivos ---")
display(df.describe().T)
```

## Fase 3: Preparación de Datos
### 3.1 Limpieza Inicial

**Detección y Tratamiento de Valores Nulos (Missing Values):**
Al revisar la estructura, detectamos datos faltantes en las columnas `artists`, `album_name` y `track_name`. Debido a que se trata de un solo registro defectuoso frente a 114,000 filas, la estrategia seleccionada es eliminar dicha fila (`dropna`). La pérdida de información es insignificante y evita sesgos.

**Eliminación de Variables Innecesarias:**
Se procedió a eliminar la columna `Unnamed: 0`. Representa únicamente un índice numérico heredado del CSV, no posee valor explicativo, por lo que su eliminación reduce la dimensionalidad.

```python
# 1. Eliminamos los valores nulos (missing values)
df = df.dropna()

# 2. Eliminamos la columna de índice innecesaria
if 'Unnamed: 0' in df.columns:
    df = df.drop(columns=['Unnamed: 0'])

print("Limpieza completada. Filas actuales:", df.shape[0])
```

### 3.2 Tratamiento de Valores Atípicos (Outliers)

**Detección y Tratamiento:**
Utilizaremos gráficos de caja (Boxplots) y el método estadístico del **Rango Intercuartílico (IQR)** para definir los límites de duración "normal". Dado que el contexto de negocio es música, las pistas extremadamente largas distorsionarán el aprendizaje de los modelos. Filtraremos y conservaremos únicamente los datos dentro de los límites superior e inferior del IQR para la variable `duration_ms`.

```python
# Aplicamos la técnica matemática IQR
Q1 = df['duration_ms'].quantile(0.25)
Q3 = df['duration_ms'].quantile(0.75)
IQR = Q3 - Q1

limite_inferior = Q1 - 1.5 * IQR
limite_superior = Q3 + 1.5 * IQR

# Filtramos el dataset
filas_antes = df.shape[0]
df = df[(df['duration_ms'] >= limite_inferior) & (df['duration_ms'] <= limite_superior)]
filas_despues = df.shape[0]

print(f"Se eliminaron {filas_antes - filas_despues} valores atípicos en la duración.")

# Visualización del resultado
plt.figure(figsize=(10, 4))
sns.boxplot(x=df['duration_ms'])
plt.title('Distribución de la Duración (duration_ms) - Limpia de Outliers')
plt.xlabel('Duración (Milisegundos)')
plt.show()
```

### 3.3 Normalización de Variables

**Justificación técnica:**
Para evitar que los futuros modelos de Machine Learning asignen un peso desproporcionado a variables por tener valores absolutos más altos (como la duración o el tempo), aplicamos **Normalización (MinMaxScaler)** mediante `scikit-learn`. Comprimiremos las variables numéricas continuas a una escala uniforme de [0, 1].

```python
# 1. Definimos qué columnas numéricas continuas vamos a normalizar
columnas_a_normalizar = [
    'duration_ms', 'loudness', 'tempo',
    'danceability', 'energy', 'speechiness',
    'acousticness', 'instrumentalness', 'liveness', 'valence'
]

# 2. Inicializamos y aplicamos la herramienta
scaler = MinMaxScaler()
df[columnas_a_normalizar] = scaler.fit_transform(df[columnas_a_normalizar])

# 3. Comprobamos el resultado (los mínimos serán 0 y máximos 1)
display(df[columnas_a_normalizar].describe().T[['min', 'max', 'mean', 'std']])
```
